# Pseudocor

Pseudocor é uma técnica para adicionar cor em uma imagem cinzenta. As cores são selecionadas em um mapa de cor pré determinado. Todos os pixels que tiverem um valor específico terão uma cor correspondente. Aplicar cores em uma imagem cinzenta ajuda a destacar algumas áreas de interesse, e facilita a visualização das mesmas. Vamos começar importando o que será usado.

```python
import cv2 as cv
import numpy as np
from auxiliar import exibirImagens
```

## Demonstração

Aqui temos um exemplo de aplicação de mapa de cor em uma imagem cinzenta. Abrimos a imagem cinzenta e aplicamos o mapa em cima dele. 

```python
imgCinza = cv.imread('images/lenaShort.jpg', cv.IMREAD_GRAYSCALE)
assert imgCinza is not None, 'Não foi possível carregar a imagem'

imgColorida = cv.applyColorMap(imgCinza, cv.COLORMAP_PINK)

exibirImagens([
    cv.cvtColor(imgCinza, cv.COLOR_GRAY2BGR),
    cv.cvtColor(imgColorida, cv.COLOR_RGB2BGR)],
    ['Cinzenta', 'Colorida'])
```

## Exercício

Vamos supor um mapa do Brasil em que cada estado tenha um tom de cinza correspondente a sua taxa de roubos de carro. Queremos aplicar uma cor para um estado específico. Vamos usar o estado de Santa Catarina como estado destaque.

```python
imgOriginal = cv.imread('images/taxaPerCapitaRouboCarros.png', cv.IMREAD_GRAYSCALE)
assert imgOriginal is not None, 'Não foi possível carregar a imagem'
valor = 120 # Valor de cinza de Santa Catarina

img = np.stack((imgOriginal, imgOriginal, imgOriginal), axis=-1)
mask = ((imgOriginal > valor - 5) & (imgOriginal < valor + 5)) # Remover ruído
img[mask] = (255,0,0)
exibirImagens([cv.cvtColor(imgOriginal, cv.COLOR_GRAY2BGR), img], ['Original', 'Colorida'])
```
